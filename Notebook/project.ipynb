### 3.Data Transformation 

**Create occupancy percentage column**

```python
df_agg_bookings.head()
```

```python
df_agg_bookings['occ_pct%'] = df_agg_bookings['successful_bookings']/df_agg_bookings['capacity']
```

```python
df_agg_bookings.head()
```

```python
df_agg_bookings['occ_pct%'] = df_agg_bookings['occ_pct%'].apply(lambda x : round(x*100,2))
df_agg_bookings.head()
```

**10.Pie chart of revenue realized per booking platform**

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = df_bookings_all.groupby("booking_platform")["revenue_realized"].sum()
colors = sns.color_palette("Blues_r", len(data))

plt.figure(figsize=(8, 8))
plt.pie(data, labels=data.index, autopct="%1.1f%%", colors=colors,
        startangle=90, wedgeprops={"edgecolor":"white", "linewidth":2})
plt.title("Revenue by Booking Platform", fontsize=15, fontweight="bold")
plt.show()
```
